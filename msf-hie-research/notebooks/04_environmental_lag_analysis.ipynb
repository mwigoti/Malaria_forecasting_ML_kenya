# 04. Environmental Lag Analysis

Assess how environmental variables and historical incidence relate to malaria risk over time.


```python
lag_columns = ["incidence_lag1", "incidence_lag2", "incidence_lag3", "incidence_lag12"]
lag_corr = df[lag_columns + ["temperature", "precipitation", "NDVI", "incidence_per_1000"]].corr()["incidence_per_1000"]
print(lag_corr.sort_values(ascending=False))

corr_df = df[["incidence_per_1000", "temperature", "precipitation", "NDVI", "ITN_coverage"]].corr().round(3)
corr_df
```

```python
for col in ["temperature", "precipitation", "NDVI"]:
    plt.figure(figsize=(10, 4))
    sns.scatterplot(data=df, x=col, y="incidence_per_1000", alpha=0.4)
    plt.title(f"{col} vs malaria incidence")
    plt.tight_layout()
    plt.show()

lag_summary = df.groupby("month")[["incidence_lag1", "temperature", "precipitation", "NDVI"]].mean().reset_index()
lag_summary
```
