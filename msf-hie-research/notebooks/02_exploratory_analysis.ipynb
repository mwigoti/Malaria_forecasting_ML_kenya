# 02. Exploratory Analysis

Inspect seasonal patterns, trends, and descriptive summaries for the malaria dataset.


```python
coverage = (
    df.groupby("county")
      .agg(start_date=("date", "min"), end_date=("date", "max"), observations=("date", "count"))
      .reset_index()
)
print("\nCounty temporal coverage:")
coverage.head()

monthly_summary = (
    df.groupby("month")["incidence_per_1000"].agg(["mean", "median", "std"]).reset_index()
)
monthly_summary

yearly_summary = (
    df.groupby("year")["incidence_per_1000"].agg(["mean", "max", "min"]).reset_index()
)
yearly_summary
```

```python
plt.figure(figsize=(12, 5))
sns.lineplot(data=df.groupby("date")["incidence_per_1000"].mean().reset_index(), x="date", y="incidence_per_1000")
plt.title("National monthly malaria incidence trend")
plt.xlabel("Date")
plt.ylabel("Incidence per 1,000")
plt.tight_layout()
plt.show()

monthly_mean = df.groupby("month")["incidence_per_1000"].mean()
plt.figure(figsize=(10, 5))
monthly_mean.plot(kind="bar", color="steelblue")
plt.title("Average incidence by month")
plt.xlabel("Month")
plt.ylabel("Incidence per 1,000")
plt.tight_layout()
plt.show()

corr = df[["incidence_per_1000", "temperature", "precipitation", "NDVI", "ITN_coverage", "population_density"]].corr()
plt.figure(figsize=(10, 8))
sns.heatmap(corr, cmap="coolwarm", annot=True, fmt=".2f")
plt.title("Correlation matrix")
plt.tight_layout()
plt.show()
```
