# 03. Spatiotemporal Analysis

Explore variation in malaria incidence across counties and time, and prepare lagged predictors for later forecasting.


```python
df["incidence_lag1"] = df.groupby("county")["incidence_per_1000"].shift(1)
df["incidence_lag2"] = df.groupby("county")["incidence_per_1000"].shift(2)
df["incidence_lag3"] = df.groupby("county")["incidence_per_1000"].shift(3)
df["incidence_lag12"] = df.groupby("county")["incidence_per_1000"].shift(12)

df["month_sin"] = np.sin(2 * np.pi * df["month"] / 12)
df["month_cos"] = np.cos(2 * np.pi * df["month"] / 12)

print(df[["county", "date", "incidence_per_1000", "incidence_lag1", "incidence_lag12"]].head(10))
```

```python
features = [
    "incidence_lag1", "incidence_lag2", "incidence_lag3", "incidence_lag12",
    "temperature", "precipitation", "NDVI", "ITN_coverage",
    "elevation", "population_density", "month_sin", "month_cos"
]
target = "incidence_per_1000"

model_df = df[["county", "date"] + features + [target]].copy()
model_df = model_df.dropna(subset=features + [target])

print(f"Rows available for modelling: {len(model_df)}")

train_df = model_df[model_df["date"].dt.year <= 2021].copy()
validation_df = model_df[model_df["date"].dt.year.between(2022, 2023)].copy()
test_df = model_df[model_df["date"].dt.year.between(2024, 2025)].copy()

print(f"Training: {len(train_df)} | Validation: {len(validation_df)} | Test: {len(test_df)}")
```
