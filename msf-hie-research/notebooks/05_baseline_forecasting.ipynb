# 05. Baseline Forecasting

Construct a strong seasonal baseline that predicts current incidence using historical incidence from the same county and month pattern.


```python
def evaluate_model(name, y_true, y_pred):
    mae = mean_absolute_error(y_true, y_pred)
    rmse = np.sqrt(mean_squared_error(y_true, y_pred))
    r2 = r2_score(y_true, y_pred)
    return {"Model": name, "MAE": mae, "RMSE": rmse, "R²": r2}

baseline_pred = test_df["incidence_lag12"].values
baseline_results = evaluate_model("Seasonal Baseline", test_df["incidence_per_1000"], baseline_pred)
baseline_results
```

```python
plt.figure(figsize=(12, 5))
actual = test_df["incidence_per_1000"].values
plt.plot(test_df["date"], actual, label="Actual", linewidth=2)
plt.plot(test_df["date"], baseline_pred, label="Seasonal Baseline", linestyle="--", linewidth=2)
plt.title("Seasonal baseline forecast versus observed incidence")
plt.xlabel("Date")
plt.ylabel("Incidence per 1,000")
plt.legend()
plt.grid(alpha=0.2)
plt.tight_layout()
plt.show()
```
